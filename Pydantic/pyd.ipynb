# 🧩 Pydantic Basics & Data Validation Guide

Welcome to the **Pydantic** beginner guide! This notebook demonstrates how to use **Pydantic** for data validation, type enforcement, and structured schema definitions in Python.

---

### 🧠 Key Concepts & Objectives

In modern Python and AI/LLM development (including LangChain and Agentic AI), **Pydantic** is essential for:
- **Type Safety & Data Parsing**: Validating input types and coercing data into Python types automatically.
- **Data Validation**: Enforcing constraints (such as minimum/maximum ranges, regex patterns, non-empty fields).
- **Structured Outputs**: Defining schema contracts for LLM function calling and structured responses.

### 📖 What This Notebook Teaches:
1. **BaseModel**: The foundation class for building Pydantic models.
2. **Field**: Specifying field-level constraints (e.g. gt for greater than, lt for less than) and metadata.
3. **Optional Type**: Handling optional fields with default values (None).
4. **Model Instantiation**: Creating and validating model objects.

## 📦 Step 1: Import Dependencies

Import BaseModel and Field from pydantic, and Optional from 	yping.
- **BaseModel**: Base class for defining schemas with type annotations.
- **Field**: Adds extra validation rules and metadata to model attributes.
- **Optional**: Defines fields that can either hold a specified type or None.

```python
from pydantic import BaseModel, Field
from typing import Optional
```

## 🏗️ Step 2: Define a Pydantic Model (Person)

Create a Person data schema inheriting from BaseModel:
- 
ame: str — Required string field.
- ge: int = Field(gt=18, lt=100) — Integer field validated to be strictly greater than 18 and less than 100.
- salary: Optional[float] = None — Optional floating-point field defaulting to None.
- city: str — Required string field.

```python
class Person(BaseModel):
    name: str
    age: int = Field(gt=18, lt=100)
    salary: Optional[float] = None
    city: str
```

## ✅ Step 3: Instantiate and Validate the Model

Create an instance of Person by supplying valid keyword arguments. Pydantic automatically validates the types and constraints upon instantiation.

```python
person = Person(name='harry potter', age=19, city='hogwards')
print(person)
```
