# ⚡ High-Speed Inference with LangChain & Groq (ChatGroq)

Welcome to the **LangChain & Groq** guide! This notebook demonstrates how to integrate Groq's high-speed LPU (Language Processing Unit) inference with LangChain using ChatGroq.

---

### 🧠 Key Concepts & Objectives

Groq provides ultra-fast inference speeds for open-weight models (such as LLaMA, Mixtral, and Gemma). Integrating Groq with LangChain allows you to achieve near-instantaneous responses for real-time AI applications.

### 📖 What This Notebook Teaches:
1. **Environment Setup**: Configuring the Groq API key and enabling LangSmith tracing.
2. **ChatGroq Integration**: Initializing and configuring Groq chat models in LangChain.
3. **Structured Messages**: Using SystemMessage and HumanMessage to define roles and guide the model's behavior.
4. **Response Handling**: Invoking the model and extracting formatted content.

## 🔑 Step 1: Environment Configuration & API Keys

Load the required API keys from your .env file:
- GROQ_API_KEY: Authenticates calls to Groq Cloud.
- LANGCHAIN_API_KEY & LANGCHAIN_TRACING_V2: Enables LangSmith tracing and observability.

```python
import os
from dotenv import load_dotenv

load_dotenv()
os.environ['LANGCHAIN_API_KEY'] = os.getenv('LANGCHAIN_API_KEY')
os.environ['LANGCHAIN_TRACING_V2'] = 'true'
groq_key = os.environ["GROQ_API_KEY"]
```

## 📦 Step 2: Import Dependencies

Import langchain and ChatGroq from langchain_groq.

```python
import langchain
from langchain_groq import ChatGroq
```

## 🚀 Step 3: Initialize the ChatGroq Model

Instantiate ChatGroq by providing the target model name (e.g., openai/gpt-oss-20b or llama3-70b-8192) and your Groq API key.

```python
llm = ChatGroq(model="openai/gpt-oss-20b", groq_api_key=groq_key)
llm
```

## 💬 Step 4: Role-Based Conversations (SystemMessage & HumanMessage)

Construct a structured conversation payload:
- **SystemMessage**: Sets the AI's persona, writing tone, and output length constraints.
- **HumanMessage**: The user's input prompt.

Invoke the ChatGroq model with the message list.

```python
from langchain_core.messages import HumanMessage, SystemMessage

message = [
    SystemMessage(content="You're an experienced english writer and philosopher, you will have to enhance the user's statement or prompt in a very peotic, expanded (50 words maximum) and philosophical manner."),
    HumanMessage(content="I am going to buy a guitar")
]

response = llm.invoke(message)
```

## 📄 Step 5: Extracting and Inspecting the Response

Access the generated text response using .content from the returned AIMessage.

```python
response.content
```
