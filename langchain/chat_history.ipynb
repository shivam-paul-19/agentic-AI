## Creating history

````python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store = {}

def get_session_history(session_id:str) -> BaseChatMessageHistory:
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]

with_history = RunnableWithMessageHistory(llm, get_session_history)
config = {"configurable":{
    "session_id":"chat_1"
}}
````

````python
response = with_history.invoke(
    [HumanMessage(content="Hi my name is shivam, I am a student in AI who is now learning about agentic AI, and that's why I am now using your LLM to learn, hope you don't mind me using you!")],
    config=config
)
````

````python
response
````

````python
config = {"configurable":{
    "session_id":"chat_2"
}}

response = with_history.invoke(
    [HumanMessage(content="What are the main 5 challenges in this field?")],
    config=config
)

response.content # it will not know the context
````
