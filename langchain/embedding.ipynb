# 🧠 LangChain OpenAI Embeddings Guide

This notebook demonstrates how to generate vector embeddings for text using LangChain and OpenAI's embedding models (`langchain_openai.OpenAIEmbeddings`).

---
### 📌 Prerequisites
1. An OpenAI API Key stored in your `.env` file (`OPENAI_API_KEY=your_key_here`).
2. Required packages installed: `langchain-openai`, `python-dotenv`.

## 🔑 Step 1: Load Environment Variables
Load the OpenAI API key from your `.env` file into the environment.

```python
import os
from dotenv import load_dotenv

load_dotenv()
os.environ['OPENAI_API_KEY'] = os.getenv('OPEN_AI_KEY')
```

## 🚀 Step 3: Import & Initialize `OpenAIEmbeddings`
Import `OpenAIEmbeddings` from `langchain_openai` and instantiate the model (defaults to `text-embedding-ada-002` or `text-embedding-3-small`).

```python
from langchain_openai import OpenAIEmbeddings

# Initialize embeddings model
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
embeddings
```

## 🔍 Step 4: Embed a Single Query (`embed_query`)
Use `embed_query()` to convert a single piece of text into a vector representation.

```python
text = "Western music theory is the study of principles and structures."
query_embedding = embeddings.embed_query(text)

query_embedding
```

## 📚 Step 5: Embed Multiple Documents (`embed_documents`)
Use `embed_documents()` to generate embeddings for a list/batch of texts or document chunks simultaneously.

```python
documents = [
    "The guitar is a popular stringed musical instrument.",
    "Acoustic guitars use hollow resonant bodies to amplify sound.",
    "Electric guitars convert string vibrations into electrical signals."
]

doc_embeddings = embeddings.embed_documents(documents)
```

```python
len(doc_embeddings[0])
```

# Ollama embedding

```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embeddings = (OllamaEmbeddings(model='nomic-embed-text'))
```

```python
emb = embeddings.embed_documents(
    [
        "Paris is the capital city of France",
        "Berlin is the capital city of Germany",
        "Tokyo is the capital city of Japan"
    ]
)

emb
```
